2.  rangsoroljuk az egyes szobákat...

```sql
SELECT sz.SZOBA_ID,
        sz.FEROHELY,
        RANK() over(ORDER by COUNT(f.SZOBA_FK) DESC, sz.ferohely asc ) as 'helyezés'
from Szoba sz left JOIN Foglalas f on sz.SZOBA_ID=f.SZOBA_FK
GROUP by sz.SZOBA_ID, sz.ferohely
order by helyezés
```
