23. Listázzuk a tanórák azonosítóját, a tanár és a tantárgy, valamint a terem nevét is! Szűrjünk csak a keddi és szerdai napokra!

```sql
SELECT o.ora_id,
        ok.nev,
        t.Tantárgynév,
        te.Terem
from Orak o JOIN Oktatok ok on o.tanar=ok.oktato_id
            JOIN Tantárgyak t on o.targy=t.Tkód
            JOIN Termek te on o.terem=te.Tkód
            JOIN Napok n on o.nap=n.nap_id
WHERE n.nap IN('kedd', 'szerda')
```

24. Készítsünk lekérdezést, amely megjeleníti azon tanár párosításokat, akiknek ugyanaz a státuszuk és beosztásuk! Jelenítsük meg a tanárok neveit, valamint a státusz és beosztás megnevezéseket!

```sql
SELECT o1.nev, o2.nev, b.beosztasnev, s.status 
FROM Oktatok o1 JOIN Oktatok o2 ON o1.oktato_id <> o2.oktato_id 
                    AND o1.beosztás = o2.beosztás
                    AND o1.status = o2.status
                JOIN Beosztasok b on o1.beosztás = b.beosztas_id 
                JOIN Statusok s on o1.status = s.skod
```

27. <span style="color: rgb(36, 41, 47); font-family: -apple-system, BlinkMacSystemFont, &quot;Segoe UI&quot;, Helvetica, Arial, sans-serif, &quot;Apple Color Emoji&quot;, &quot;Segoe UI Emoji&quot;; font-size: 16px;">Listázzuk nyelvenkénti, azon belül stílusonkénti bontásban, hogy mennyi a dvd-k legkisebb és legnagyobb ára!&nbsp;</span> 
    
    <span style="font-size: 16px;">a. Az oszlopok neve legyen 'Nyelv', 'Stílus', 'MinÁr' és 'MaxÁr'!<br>b.&nbsp;</span>  A listából hagyjuk ki azokat a sorokat, ahol a nyelv vagy a stílus nincs megadva!  
    c. Szintén hagyjuk ki azokat a csoportokat, ahol a csoport elemszáma 3-nál kisebb!

```sql
SELECT n.lang,
        stilus,
        MIN(nettoar) as 'Minár',
        MAX(nettoar) as 'Maxár'
from dvd d JOIN nyelvek n on d.lang=n.id
WHERE n.lang IS Not NULL and stilus is not NULL
GROUP by n.lang, stilus
HAVING COUNT(*)>=3
```

30. <span style="font-weight: 600; color: rgb(36, 41, 47); font-family: -apple-system, BlinkMacSystemFont, &quot;Segoe UI&quot;, Helvetica, Arial, sans-serif, &quot;Apple Color Emoji&quot;, &quot;Segoe UI Emoji&quot;; font-size: 16px;">A dvd-ket csoportosíthatjuk aszerint, hogy milyen hosszú a címük. 42. <span style="font-weight: 600; color: rgb(36, 41, 47); font-family: -apple-system, BlinkMacSystemFont, &quot;Segoe UI&quot;, Helvetica, Arial, sans-serif, &quot;Apple Color Emoji&quot;, &quot;Segoe UI Emoji&quot;; font-size: 16px;">Készítsünk listát arról, hogy melyik tanárnak hány órája van a (munka) hét elején (hétfőn), hét közben (kedd, szerda és csütörtök), illetve a hét végén (péntek)!</span>
    
    1. Jelenítsük meg a tanár azonosítóját, nevét, az időszakot (hét eleje - hét közben - hét vége), valamint az óraszámot!
    2. A három megjelölt időszakot külön kérdezzük le, majd fűzzük össze a lekérdezések eredményeit!
    3. Rendezzük a listát az oktató neve, azon belül időszak szerint!

```sql
SELECT okt.oktato_id,
       okt.nev,
       CASE 
            WHEN n.nap = N'Hétfő' THEN 'Hét eleje'
            WHEN n.nap IN ('Kedd', 'Szerda', 'Csütörtök') THEN 'hét közepe'
            ELSE 'hét vége'
       END AS 'Időszak',
       COUNT(*) as 'db'
FROM Oktatok okt JOIN Orak o ON okt.oktato_id = o.tanar
                 JOIN Napok n ON o.nap = n.nap_id
GROUP BY okt.oktato_id, 
         okt.nev,
         CASE 
            WHEN n.nap = N'Hétfő' THEN 'Hét eleje'
            WHEN n.nap IN ('Kedd', 'Szerda', 'Csütörtök') THEN 'hét közepe'
            ELSE 'hét vége'
         END
ORDER BY okt.oktato_id, okt.nev, 'Időszak'
```

44.  _Készítsünk listát az oktatók adatairól!_

1. Az oktatók titulusa és neve összefűzve jelenjen meg, közöttük szóközzel!
2. Ha valakinek nincs titulusa, akkor csak a neve jelenjen meg, szóköz nélkül!

4. Az oszlopokat nevezzük el értelemszerűen!

```sql
SELECT oktato_id,
        case 
        when titulus is null then nev
        when titulus is not null then titulus+' '+nev
        End as 'Név',
        beosztás as 'Beosztás',
        [status] as 'Státusz'
from Oktatok
ORDER BY beosztás, [status] ASC
```

45. Készítsünk listát, amely megmutatja, hogy összesen hány tantárgy van kreditenkénti csoportosításban!
    
    A listából hagyjuk ki azokat a tételeket, ahol ez a darabszám <span style="color: rgb(9, 136, 90);">1</span>!
    
    A Folyami gáttan tantárgyat ne vegyük figyelembe!
    
    A listát rendezzük darabszám szerint csökkenő sorrendbe!
    
    Az oszlopokat nevezzük el értelemszerűen!

```sql
SELECT Kredit,
        COUNT(Tkód) as 'Tantárgyak száma'
from Tantárgyak
WHERE Tantárgynév <> 'Folyami gáttan'
GROUP BY Kredit
HAVING COUNT(Tkód)>1
ORDER BY [Tantárgyak száma] DESC
```

46. _Készítsünk listát az oktatók nevéről, a beosztásuk alapján elvárt és a tényleges óraszámukról!_

    2. Szűrjük a listát azokra az oktatókra, ahol az elvárt és a tényleges óraszám azonos

```sql
SELECT ok.nev as 'Oktató neve',
        b.elvart as 'elvárt óraszám',
        COUNT(*) as 'Tényleges óraszám'
from Oktatok ok JOIN Beosztasok b on ok.beosztás=b.beosztas_id
                JOIN Orak o on ok.oktato_id=o.tanar
GROUP BY ok.nev, b.elvart
HAVING b.elvart=COUNT(*)
```

48. _Hány órájuk van az óraadóknak napi bontásban?_

1. Csak a napok nevét és az óraszámot jelenítsük meg!
2. Rendezzük a listát óraszám szerint növekvő sorrendbe!

```sql
SELECT n.nap,
        COUNT(o.ora_id) as 'db'
from Orak o JOIN Oktatok ok on o.tanar=ok.oktato_id
            JOIN Beosztasok b on ok.beosztás=b.beosztas_id
            JOIN Napok n on o.nap=n.nap_id
WHERE b.beosztasnev='Óraadó'
GROUP BY n.nap
ORDER BY db ASC
```
