````sql
SELECT case GROUPING_ID(YEAR(r.REND_DATUM)) -- a végösszeg jelölése
        when 1 then 'Összesen'
        else CAST(YEAR(r.REND_DATUM) as nvarchar(4))
        end AS 'Év',
        SUM(rt.mennyiseg*rt.egysegar)
from Rendeles r JOIN Rendeles_tetel rt on r.SORSZAM=rt.SORSZAM
where r.[LOGIN]='norbert2'
GROUP BY rollup(YEAR(r.REND_DATUM))
````

1. Listázzuk az egyes csoportokat, és a csoportokba tartozó termékek darabszámát!
2. A lista jelenítse meg a végösszeget is!

````sql
SELECT case 
        when LISTAAR>5000 then 'drága'
        when LISTAAR<3000 then 'olcsó'
        else 'közepes' 
        end as 'árkategória',
        COUNT(TERMEKKOD)
From Termek
GROUP BY --olcso, kozepes vagy draga
        rollup(case 
        when LISTAAR>5000 then 'drága'
        when LISTAAR<3000 then 'olcsó'
        else 'közepes'
        end)
````

2. Ha nincs előző vagy következő ügyfél, akkor a 'Nincs' jelenjen meg!

````sql
select *,
        --elozo ugyfel neve is kell
        lag(NEV, 1, 'Nincs') over(order by NEV) as 'előtte',
        LEAD(NEV, 1, 'Nincs') over(order by NEV) as 'utána'
from Ugyfel
order by NEV
````

2. A sorszám oszlop neve legyen Azonosító

````sql
SELECT
        --sorszám
        cast(ROW_NUMBER() over(PARTITION by [LOGIN], YEAR(REND_DATUM) order by SORSZAM) as nvarchar(10))
        +'_'+CAST(YEAR(REND_DATUM) as nvarchar(4)) + '_'+ [LOGIN] as 'azonosító',
        *
from Rendeles
ORDER by [LOGIN], REND_DATUM
````

1. Vegyünk fel egy új oszlopot Mozgóátlag néven, amely minden esetben az aktuális termék, az előző, és a következő termék átlagárát tartalmazza!
2. A mozgóátlagot kerekítsük két tizedesre!

````sql
SELECT TERMEKKOD,
        MEGNEVEZES,
        LISTAAR,
        --mozgoatlag kiszamitasa kell
        --(elozo+aktualis+kovetkezo)/3
        AVG(LISTAAR) OVER(order by LISTAAR rows BETWEEN 1 preceding and 1 following) as 'mozgóátlag'
from Termek
ORDER by LISTAAR
````
